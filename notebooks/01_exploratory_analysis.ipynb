```python
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
p = Path('../data/processed/daily_sales.csv')
if not p.exists(): p = Path('data/processed/daily_sales.csv')
df = pd.read_csv(p, parse_dates=['date'])
df.head()
```

```python
print('Rows',len(df),'Missing:',df.isna().sum())
print('Duplicates:',df.duplicated(['date','store_id','product_id']).sum())
```

```python
daily=df.groupby('date').units_sold.sum()
daily.rolling(7).mean().plot(figsize=(11,4),title='Synthetic weekly trend')
plt.show()
df.groupby('product_id').units_sold.sum().plot.barh()
plt.show()
```

**Disclaimer:** insights in this notebook reflect simulation settings, not real retailer behavior.
